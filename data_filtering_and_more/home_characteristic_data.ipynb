```python

df = df_raw.copy()
df['missing_count'] = df[cols].isnull().sum(axis=1)

df = df.sort_values('missing_count', ascending=False)

n = len(df)
height_per_bar = 0.3
total_height = (n) * height_per_bar   # +2 inches padding

plt.figure(figsize=(10, total_height))
plt.barh(df['Property_ID'].astype(str), df['missing_count'], edgecolor='white')

plt.barh(
    df['Property_ID'].astype(str),
    df['missing_count'],
    edgecolor='black'
)
plt.xlabel('Number of Missing Fields')
plt.ylabel('Property_ID')
plt.title('Missing Data Count per Property')
plt.tight_layout()
plt.ylim(-0.5, len(df) - 0.5)
plt.show()
```
